```python
#converting date columns to dates
date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"

]
for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")
```

```python
# replace the missing values 
df['Customer Zipcode'] = df['Customer Zipcode'].fillna(0)
df['Order Zipcode'] = df['Order Zipcode'].fillna(0)

# create customer full name column 
df['Customer Full Name'] = (
    df['Customer Fname'].astype(str)
    + " "
    + df['Customer Lname'].astype(str)
)
df.columns
```

```python
# droping unnecessary columns 
dfnew=df.drop(['Customer Email','Product Status','Customer Password','Customer Street','Customer Fname','Customer Lname',
           'Product Description','Product Image','Order Zipcode','Customer Zipcode'],axis=1)
dfnew.shape
```

```python
dfnew.columns = [col.lower().replace(' ', '_') for col in dfnew.columns]
dfnew.rename(columns=lambda x: x.replace("(", "").replace(")", ""), inplace=True)
dfnew.columns
```

```python
# create a year and a month column for better more specified analysis

dfnew["order_year"] = dfnew["order_date_dateorders"].dt.year
dfnew["order_month"] = dfnew["order_date_dateorders"].dt.month
dfnew.head(3)
```

```python
order_items = dfnew[['order_item_id',
    'order_id',
    'product_card_id',
    'order_item_discount',
    'order_item_discount_rate',
    'order_item_product_price',
    'order_item_profit_ratio',
    'order_item_quantity',
    'order_item_total',
    'sales',
    'order_profit_per_order']]
display(order_items)
```
